```python
#3rd element and 3rd element from last of a tuple
t=(10,20,30,40,50)
print("3rd element:",t[2])
print("3rd element from last:",t[-3])
```

```python
#comma separated numbers from user and generate a tuple find average of all items in the tuple
numbers=input("enter comma separated numbers:")
t=tuple(map(int,numbers.split(",")))
print("tuple:",t)
average=sum(t)/len(t)
print("average:",average)
```
