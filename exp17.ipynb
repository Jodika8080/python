```python
#read a file line by line and store
l=[]
file=input("enter the file name:")
f=open(file,"r")
for line in f.readlines():
    l.append(line)
print("lines in the list:",l)
print("no of lines in the list:",len(l))
f.close()
```

```python
#copy odd lines of one file to another
file1=input("enter the first file name:")
file2=input("enter the second file name:")
i=0
with open(file1,"r")as f1:
     with open(file2,"w")as f2:
         for line in f1.readlines():
             i=i+1
             if i%2==0:
                   continue
             f2.write(line)
print("odd lines copied succccessfully.")
with open(file2,"r")as f2:
    print("contents(copied odd lines)of second file:")
    print(f2.read())
```
