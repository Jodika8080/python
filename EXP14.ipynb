```python
%%writefile graphics/Circle.py
import math
def Area(r):
    return math.pi * r * r
def Perimeter(r):
    return 2 * math.pi * r
    
```

```python
%%writefile graphics/rectangle.py
def Area(l,b):
     return l*b
def Perimeter(l,b):
     return 2*(l+b)
```

```python
%%writefile graphics/graphics3D/cuboid.py
def area(l,b,h):
    return 2*(l*b+l*h+b*h)
def perimeter(l,b,h):
    return 4*(l+b+h)
    
```

```python
%%writefile graphics/graphics3D/sphere.py
import math
def Area(r):
    return 4*math.pi*r*r
def Perimeter(r):
    return 2*math.pi*r
    
```
