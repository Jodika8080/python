```python
#n+nn+nnn
n=int(input("enter a value of n:"))
nn=n*11
nnn=n*111
sum=int(n)+int(nn)+int(nnn)
print(n,"+",nn,"+",nnn,"=",sum)
```

```python
#radius of a circle
import math
r=float(input("Enter the radius:"))
area = math.pi*r*r
print("area of the circle=",f"{area:.2f}")
```

```python
#find biggest number
a=int(input("enter first number:"))
b=int(input("enter second number:"))
c=int(input("enter third number:"))
biggest=max(a,b,c)
print("biggest number =",biggest)
```
