```python
#accept a list of words and return the length of the largest word
def longest_word(words):
    longest=max(len(word) for word in words)
    return longest
words=input("enter the words:").split()
print("the words are:",words)
length=longest_word(words)
print("length of the longest word is:",length)
```

```python
#construct a pattern using nested loop
def pattern(n):
    for i in range(1,n+1):
        for j in range(i):
            print("*", end=" ")
        print()
    for i in range(n-1,0,-1):
        for j in range(i):
            print("*",end=" ")
        print()
n=int(input("enter the limit:"))
print("the number:",n)
pattern(n)
```

```python
#Generate all factors of a number.
def factors(n):
    print("Factors are:")
    for i in range(1,n+1):
        if n%i==0:
            print(i, end=" ")
n = int(input("Enter the number: "))
print("The number:", n)
factors(n)
```

```python
#lambda functions to find the area of square,rectangle,and triangle.
square=lambda a:a*a
rectangle=lambda l,b:l*b
triangle=lambda b,h:0.5*b*h
print("Area of square:",square(4))
print("Area of rectangle:",rectangle(5,3))
print("Area of triangle:",triangle(6,4))
```
