```python
#generate a positive list of numbers
numbers=[10,-5,20,-8,15,0]
positive=[x for x in numbers if x>0]
print("positive numbers:",positive)
```

```python
#square of N numbers
n=int(input("enter the number of elements:"))
numbers=[int(input())for i in range(n)]
squares=[x*x for x in numbers]
print("squares:",squares)
```

```python
#form a list of vowels from given word
word=input("enter a word:")
vowels=[ch for ch in word if ch.lower() in 'aeiou']
print("vowels:",vowels)
```

```python
#list orginal value of each character in word
word=input("enter a word:")
values=[ord(ch) for ch in word]
print("orginal values:",values)
```
