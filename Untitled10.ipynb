```python
#without list comprehension
squares=[]
for x in range(5):
    squares.append(x**2)
print(squares)
squares=[x**2 for x in range(5)]
print(squares)
```

```python
#string operation
words=["apple","banana","cherry"]
upper=[w.upper()for w in words if "a" in w]
print(upper)
```

```python
#flatten a matrix
matrix=[[1,2],[3,4],[5,6]]
flattened=[num for row in matrix for num in row]
print(flattened)
```

```python
nums=[x for x in range(20) if x % 2 ==0 if x %3==0]
print(nums)
```

```python
def square(x):
  return x*x
result=[square(i) for i in range(6)]
print(result)
```
