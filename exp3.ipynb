```python
#exchange the first and last char
s=input("enter a string:")
new_string=s[-1] + s[1:-1] + s[0]
print("new string:",new_string)
```

```python
#first 2 and last char
s = input("enter a string:")
new_string = s[:2]+s[-2:]
print("new string:",new_string)
```

```python
#swap character at 1st position
s1=input("enter first string:")
s2=input("enter second string:")
new_string=s2[0]+s1[1:]+" "+s1[0]+s2[1:]
print("after swap:",new_string)
```

```python
# replace all occurance of the fist char except the first one
s=input("enter a string:")
new_string=s[0]+s[1:].replace(s[0],"$")
print("new_string:",new_string)
```
