```python
#add "ing" or "ly" to a string
s=input("enter a string:")
if s.endswith("ing"):
   s=s+"ly"
else:
    s=s+"ing"
print(s)
```

```python
#check the string is palindrom or not
s=input("enter a string :")
if s==s[::-1]:
    print(s," is palindrom")
else:
    print(s," is not a palindrom")
```
